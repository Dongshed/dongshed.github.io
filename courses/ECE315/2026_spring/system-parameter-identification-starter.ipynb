The underlying physical system is a series RC circuit. The resistance $R$ and the capacitance $C$ are unknown. The input $x(t)$ is the voltage $v(t)$ and the output $y(t)$ is the voltage on the capacitor $v_C(t)$. According to Kirchhoff's voltage law, 

$$Ri(t) + v_C(t) = v(t).$$

Since the current $i(t) = C\frac{dv_C(t)}{dt}$, we describe the RC circuit as the following differential equation:

$$RC\frac{dv_C(t)}{dt} + v_C(t) = v(t).$$

Equivalently, 

$$RC\frac{d y(t)}{dt} + y(t) = x(t).$$

Assume that the system has initial state, i.e., $v_C(0)=0$. The time constant of a RC circuit is given by $\tau = RC$. We next observe a few input and output signals to determine the time constant $\tau$.
